```python
def simple_reflex_agent(room, status):
    if status[room] == "Dirty":
        print("Room", room, "is Dirty")
        print("Action: SUCK")
        status[room] = "Clean"
    else:
        print("Room", room, "is Clean")

        if room == "A":
            print("Action: MOVE RIGHT")
            room = "B"
        else:
            print("Action: MOVE LEFT")
            room = "A"

    return room


def run_simple_reflex_agent():
    print("\n--- SIMPLE REFLEX AGENT ---")

    A = input("Enter Room A status (Dirty/Clean): ")
    B = input("Enter Room B status (Dirty/Clean): ")

    status = {"A": A, "B": B}

    room = input("Enter starting room (A/B): ")

    for i in range(4):
        room = simple_reflex_agent(room, status)
        print("Status:", status)


def goal_based_agent(room, status):
    if status["A"] == "Clean" and status["B"] == "Clean":
        print("Goal achieved!")
        return room

    if status[room] == "Dirty":
        print("Room", room, "is Dirty")
        print("Action: SUCK")
        status[room] = "Clean"

    elif room == "A":
        print("Action: MOVE RIGHT")
        room = "B"

    else:
        print("Action: MOVE LEFT")
        room = "A"

    return room


def run_goal_based_agent():
    print("\n--- GOAL-BASED AGENT ---")

    A = input("Enter Room A status (Dirty/Clean): ")
    B = input("Enter Room B status (Dirty/Clean): ")

    status = {"A": A, "B": B}

    room = input("Enter starting room (A/B): ")

    while status["A"] != "Clean" or status["B"] != "Clean":
        room = goal_based_agent(room, status)

    print("Final Status:", status)
    print("Goal achieved!")


run_simple_reflex_agent()
run_goal_based_agent()
```
